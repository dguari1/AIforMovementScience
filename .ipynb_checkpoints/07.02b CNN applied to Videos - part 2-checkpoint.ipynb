# CNN Applied to videos - part 2

In this notebook, we will apply out object localization model to track the movement of weights during a weight lifting exercise. 

What you will learn in this notebook:
- How to apply a CNN to each video frame
- How to extract object position over time
- How to track motion
- Why signals from video can be noisy

```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from PIL import Image
```

```python
img = Image.open(r'data/lifter.png') #load the image and convert it to RGB format
plt.imshow(img) #display the image using matplotlib
```

We will use our model to check what can be detected in this image

```python
model = YOLO("models/yolo26s.pt")
results = model(img)
boxes = results[0].boxes.xyxy.cpu().numpy() #get the bounding boxes in xyxy format and convert to numpy array
classes = results[0].boxes.cls.cpu().numpy() #get the class labels of the detected objects and convert to numpy array
plt.imshow(img)
for idx, box in enumerate(boxes):
    x1, y1, x2, y2 = box
    plt.gca().add_patch(plt.Rectangle((x1, y1), x2-x1, y2-y1, fill=False, color='red', linewidth=2)) #draw the bounding box on the image
    plt.text(x1, y1, f'{model.names[int(classes[idx])]}', color='red', fontsize=12) #add the class label to the bounding box
plt.show()
```

The model did a great job finding people in the image, but the model did not find the weights... 

It turns out that the weights are not part of the COCO categories...

We can now do this for the full video. 

```python
from IPython.display import Video, display

video_path = r"data/lift_one.mp4"

cap = cv2.VideoCapture(video_path)
if not cap.isOpened():
    raise FileNotFoundError(f"Could not open video: {video_path}")
cap.release()

display(Video(video_path, embed=True))
```

The objective now is to to an automatic process to localize the barbell in each video frame and keep track of its position. For this, we will:
1. Detect object in each frame
2. Extract position
3. Store positions over time
4. Smooth signal
5. Analyze motion

```python
barbellmodel = YOLO("models/barbelldetector.pt")
video = cv2.VideoCapture(r'data/lift_one.mp4')

target_names = {"barbell"}
name_to_id = {name: idx for idx, name in barbellmodel .names.items()}
target_ids = [name_to_id[name] for name in target_names if name in name_to_id]

positions = []

while True:
    valid, frame = video.read() 
    

    if not valid:
        break
    #crop the frame to the region of interest (where the barbell is expected to be)
    frame = frame[:,600:1500,:]
    r = barbellmodel.track(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB), persist=True, classes=target_ids, verbose=False)[0]

    frame_data = {
        "frame": video.get(cv2.CAP_PROP_POS_FRAMES),
        "time": video.get(cv2.CAP_PROP_POS_MSEC) / 1000,
        "barbell_x": np.nan, "barbell_y": np.nan, 
    }


    if r.boxes is not None and len(r.boxes) > 0:
        cls = r.boxes.cls.cpu().numpy().astype(int)
        conf = r.boxes.conf.cpu().numpy()
        xyxy = r.boxes.xyxy.cpu().numpy()  # x1, y1, x2, y2

        for label in ["barbell"]:
            cid = name_to_id[label]
            idx = np.where(cls == cid)[0]
            if idx.size > 0:
                best = idx[np.argmax(conf[idx])]  # highest confidence detection
                x1, y1, x2, y2 = xyxy[best]
                # Calculate center and dimensions
                x = (x1 + x2) / 2
                y = (y1 + y2) / 2
                frame_data[f"{label}_x"] = float(x)
                frame_data[f"{label}_y"] = float(y)
    positions.append(frame_data)
video.release() # release the video when done to free up resources
```

```python
#plot the barbell position over time
df = pd.DataFrame(positions)
plt.figure(figsize=(10, 6))
plt.plot(df['time'], df['barbell_x'], label='Barbell X Position')
plt.plot(df['time'], df['barbell_y'], label='Barbell Y Position')
plt.xlabel('Time (s)')
plt.ylabel('Position (pixels)')
plt.title('Barbell Position Over Time')
plt.legend()
plt.show()
```

There is a large jump in the position due to miss localization for one frame. Detection may fail because:
- occlusion
- motion blur
- lighting
- model limitations

We can smooth the signal to make sure that these harsh transitions are removed

```python
# Rate-limit + smooth to avoid harsh frame-to-frame position jumps
def smooth_signal_with_max_step(signal, max_step, window=5):
    arr = signal.to_numpy(dtype=float)
    out = arr.copy()

    # Find first valid value
    valid_idx = np.where(np.isfinite(arr))[0]
    if len(valid_idx) == 0:
        return pd.Series(out, index=signal.index)

    start = valid_idx[0]
    for i in range(start + 1, len(arr)):
        if not np.isfinite(arr[i]):
            out[i] = out[i - 1]
            continue
        delta = arr[i] - out[i - 1]
        out[i] = out[i - 1] + np.clip(delta, -max_step, max_step)

    # Light moving-average smoothing
    out = pd.Series(out, index=signal.index).rolling(window=window, center=True, min_periods=1).mean()
    return out

# Adaptive thresholds from observed motion
max_step_x = float(df["barbell_x"].diff().abs().quantile(0.95))
max_step_y = float(df["barbell_y"].diff().abs().quantile(0.95))

df["barbell_x_filt"] = smooth_signal_with_max_step(df["barbell_x"], max_step=max_step_x, window=5)
df["barbell_y_filt"] = smooth_signal_with_max_step(df["barbell_y"], max_step=max_step_y, window=5)

# Compare original vs filtered
plt.figure(figsize=(11, 5))
plt.plot(df["time"], df["barbell_x"], alpha=0.35, label="barbell_x (raw)")
plt.plot(df["time"], df["barbell_x_filt"], linewidth=2, label="barbell_x (filtered)")
plt.plot(df["time"], df["barbell_y"], alpha=0.35, label="barbell_y (raw)")
plt.plot(df["time"], df["barbell_y_filt"], linewidth=2, label="barbell_y (filtered)")
plt.xlabel("Time (s)")
plt.ylabel("Position (pixels)")
plt.title("Barbell position signals (raw vs filtered)")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

We can also plot the X,Y position of the barbell and see how it moves in space

```python
# Plot barbell trajectory in X-Y space
plt.figure(figsize=(7, 7))
plt.plot(df["barbell_x_filt"], df["barbell_y_filt"], linewidth=2, label="Filtered trajectory")

# Mark start/end points (filtered)
plt.scatter(df["barbell_x_filt"].iloc[0], df["barbell_y_filt"].iloc[0], c="green", s=60, label="Start")
plt.scatter(df["barbell_x_filt"].iloc[-1], df["barbell_y_filt"].iloc[-1], c="red", s=60, label="End")

plt.xlabel("X position (pixels)")
plt.ylabel("Y position (pixels)")
plt.title("Barbell X-Y Position")
plt.gca().invert_yaxis()  # image coordinates: y increases downward
plt.axis("equal")
plt.grid(alpha=0.2)
plt.legend()
plt.show()
```
