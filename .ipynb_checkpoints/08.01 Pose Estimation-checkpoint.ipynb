# Introduction to Pose Estimation

Pose estimation is a computer vision task that aims to detect and track key points (landmarks) on objects. Pose estimation is typically applied to the human body; however, it can be extended to localize identifiable landmarks in any object or animal. 

These keypoints represent identifiable locations such as:
- joints (e.g., shoulders, elbows, knees)
- extremities (e.g., hands, feet)
- facial landmarks

Pose estimation is widely used in:
- human movement analysis
- sports science
- rehabilitation
- neuroscience
- human-computer interaction
- animation and gaming

# What is Pose Estimation?

Pose estimation models take an image as input and output the coordinates of keypoints.

## Type of Models

There are two main types of models:
### 1. Regression models
- These models receive an image as input and their output is the prediction of the *(x,y)* position of the different keypoints. These models are usually fast but may lack in accuracy. 

### 2. Heatmaps models
- These models receive an image an input and output is a set of heatmaps with the likely *(x,y)* position of the different keypoints. 
<div style="text-align:center">
<img src="images/heatmap-based-models.png" alt="plot" width="400" />
</div>

There is one heatmap per keypoint, and the position of each kepoint is represented by a 2D Gaussian function. The mean represents the expected keypoint position and the standard deviation is a measure of confidence. Occlusions can also be easily handled by producing empty heatmaps.


## Tasks

There are three main task performed by pose estimation models:

### 1. 2D Pose Estimation
- Outputs (x, y) pixel coordinates

<div style="text-align:center">
<img src="images/COCOPose.webp" alt="plot" width="400" />
</div>

### 2. 3D Pose Estimation
- Outputs (x, y, z) coordinates
- Estimated depth information

<div style="text-align:center">
<img src="images/3dpose.gif" alt="plot" width="600" />
</div>

**3D Pose Estimation** can be computed in three different ways (in order of accuracy):

a - Use a depth sensor to measure the distance from the different keypoints to the camera. 

b -  Use multiple cameras to reconstruct the 3D scene using triangulation.

c - Use a depth model to estimate the distance from the different keypoints to the camera. 

### 3. Volumetric Pose Estimation
- Volumetric models use the output of 3D pose estimation models to fit a deformable mesh to a 2D image. 
- These meshes usually have thousands of vertex that are adjusted to fit the estimated 3D landmarks and conform to a specific shape (i.e. a human shape). 

<div style="text-align:center">
<img src="images/mesh.png" alt="plot" width="600" />
</div>

## Common Pose Estimation Problems

Pose estimation models can focus on different body parts:

### Full Body Pose
- Detects major joints (e.g., shoulders, hips, knees)
- Most models estimate 17–33 keypoints; however, many models estimate more keypoints. 

### Hand Pose
- Detects the position of different finger joints
- Most models estimate the position of 21 keypoints per hand (finger joints + tips and wrist). 

### Face Pose
- Detects facial landmarks. (eyes, nose, mouth, jawline)
- Most models detect a few keypoints (either 38 or 68) around the eyes, nose, mouth, and jawline. However, some models include hundreds of keypoints.

### Whole Body Pose
- These models combine body, hands, and face.
- Used in detailed motion analysis and animation

## Pose Estimation Datasets

Pose estimation models are trained on labeled datasets where keypoints are manually annotated.

### Common Datasets

#### COCO Pose (Common Objects in Context)
- 17 keypoints for human body
- One of the most widely used datasets

#### MPII Human Pose
- Uses 16 keypoints, it is mostly focuses on human activities

- Captured in controlled lab environments with actors performing multiple movements. Data was captured with an array of cameras.

## Popular Pose Estimation Models

Some models are designed for specific use cases:

### YOLO Pose
- Detects 2D keypoints
- Designed for high processing speed while maintaining high accuracy. 
- Trained on the COCO Pose dataset, but can be fine-tuned with any other dataset. 

### MediaPipe
- Detects 3D keypoints
- Designed for high processing speed while maintaining high accuracy. 
- Trained on the MediaPipe dataset. It is a closed model, it cannot be modified.  

#### MediaPipe Pose
- It is a lightweight and accurate model that Detects 3D body keypoints
<div style="text-align:center">
<img src="images/mediapipePose.png" alt="plot" width="400" />
</div>

#### MediaPipeHands
 - It is a lightweight and accurate model that Detects 3D hand keypoints

<div style="text-align:center">
<img src="images/mediapipeHands.jpg" alt="plot" width="400" />
</div>

#### MediaPipe Face
- It is a lightweight and accurate model that Detects 3D face keypoints
<div style="text-align:center">
<img src="images/mediapipeFace.gif" alt="plot" width="200" />
</div>

### HRNet
HRNet are a modern family of pose estimation models designed to work with high-resolution images. Most models start by downsampling the image to speed up the process, loosing important spatial information. In contrast, HRNet works with the full resolution image preserving this information and providing more accurate results. 

HRNet models are avaliable as part of the MMPose library.

### Whole Body
- OpenPose and AlphaPose are a popular models that combines body, hands, and face.

<div style="text-align:center">
<img src="images/fullbody.webp" alt="plot" width="400" />
</div>

### WHAM
- Allows you to reconstruct 3D mesh from single images, designed for motion consistency between frames. 

### Animals
- Pose estimation models have also been extended to other application, including animals. A few databases for animal pose estimation have been developed over the years and multiple models are available in libraries such as MMPose.

```python
from ultralytics import YOLO
import cv2
import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import IntSlider, Play, HBox, VBox, jslink, interactive_output
from IPython.display import display


edges = [
    (5, 7), (7, 9),     # left arm
    (6, 8), (8, 10),    # right arm
    (5, 6),             # shoulders
    (5, 11), (6, 12),   # torso sides
    (11, 12),           # hips
    (11, 13), (13, 15), # left leg
    (12, 14), (14, 16), # right leg
    (0, 5), (0, 6)      # head to shoulders
]
```

```python
video_path = "data/lift_one.mp4"
model = YOLO("models/yolo26s-pose.pt")  # COCO 17-keypoint pose model

#YOLO pose can run the models directly from the video!
# Run pose tracking
results = model.track(
    source=video_path,
    stream=True,
    tracker="bytetrack.yaml",
    persist=True,
    verbose=False
)
#get info from the video to make the plot
cap = cv2.VideoCapture(video_path)
w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
cap.release()

#extracting keypoints from the results
frame_ids = []
pose_seq = []  # (num_frames_detected, 17, 2)

for i, r in enumerate(results):
    if r.keypoints is None or len(r.keypoints.xy) == 0:
        continue

    # Pick the highest-confidence person in frame
    kpts = r.keypoints.xy.cpu().numpy()  # (n_people, 17, 2)
    conf = r.boxes.conf.cpu().numpy() if r.boxes is not None else np.ones(len(kpts))
    person_idx = int(np.argmax(conf))

    frame_ids.append(i)
    pose_seq.append(kpts[person_idx])

#plotting the keypoints and skeleton connections for a given frame index
def plot_pose_frame(frame_idx):
    kpts_frame = pose_seq[frame_idx]

    fig, ax = plt.subplots(figsize=(8, 5))
    ax.scatter(kpts_frame[:, 0], kpts_frame[:, 1], s=30, c="red")

    # Draw skeleton connections
    for a, b in edges:
        ax.plot(
            [kpts_frame[a, 0], kpts_frame[b, 0]],
            [kpts_frame[a, 1], kpts_frame[b, 1]],
            color="cyan",
            linewidth=2
        )

    ax.set_xlim(0, w)
    ax.set_ylim(h, 0)  # image coordinates
    ax.set_aspect("equal", adjustable="box")
    ax.set_title(f"Pose keypoints - frame {frame_ids[frame_idx]} (index {frame_idx})")
    ax.set_xlabel("X (pixels)")
    ax.set_ylabel("Y (pixels)")
    ax.grid(alpha=0.2)
    plt.show()

slider = IntSlider(value=0, min=0, max=len(pose_seq) - 1, step=1, description="Frame")
play = Play(value=0, min=0, max=len(pose_seq) - 1, step=1, interval=80, description="Press play")
jslink((play, "value"), (slider, "value"))

ui = HBox([play, slider])
out = interactive_output(plot_pose_frame, {"frame_idx": slider})
display(VBox([ui, out]))
```
