All three files come from the Gait in Parkinson's Disease database (PhysioNet, <https://physionet.org/content/gaitpdb/>). They are already in the course files, in `lectures/data`. The setup cell stored that path in `DATA`.

**`SiCo01_01.txt` and `SiPt02_01.txt` — two raw force recordings.**
Each subject walked for about two minutes with eight force sensors under each foot. The sensors measure the vertical force between the foot and the floor, in Newtons. The signal is sampled at 100 Hz. Each file is a plain table of numbers with 19 columns:

| 18 | total force under the right foot |

`SiCo01` is a control subject: 53 years old, 56 kg.
`SiPt02` has Parkinson's disease: 77 years old, 67 kg.

**`gait_measures_PD_controls_ALL.csv` — 306 rows, one per walking recording.**
Eighteen gait measures already computed from recordings like the two above: walking speed, cadence (steps per minute), step length, stride and stance times, and the coefficient of variation (`cv_`) of each timing measure. The `class` column is `0` for a control subject and `1` for a subject with Parkinson's disease. The `ID` column names the subject who produced the recording.

**`demographics_Parkinsons.csv` — age, height, weight and clinical scores.**
