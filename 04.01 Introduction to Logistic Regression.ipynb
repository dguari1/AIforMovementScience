## Classification

In this notebook, we will introduce the concept of classification and learn how to apply our knowledge of linear regression to address this new machine learning challenge.

In linear regression, we found a relationship between independent variables (inputs) and dependent variables (outputs), both of which are continuous variables. 

Similarly, in classification, we are interested in finding a relationship between independent variables (features) and a dependent variable (classes). However, in this case, the features are continuous or discrete variables, while the classes are discrete. 


| feature 1 | feature 2 | Class|

|0.25 | 0.13| 0|
|0.45 | 0.23| 0|
|0.12 | 0.98| 1|
|0.35 | 0.16| 0|
|0.05 | 0.74| 1|
|0.36 | 0.08| 0|


The objective of the classification problem is to identify the relationship between the features and the classes, such that when new features are available, it is possible to predict the class.
