There are some restrictions on the names variables can take. 
- Variables can only contain alphanumeric characters (letters and numbers) as well as underscore,
- the first character of a variable name must be a letter or underscore,
- Variable names with spaces are not permitted, and 
- The variable names are case-sensitive (e.g., x and X will be considered different variables).

Python allows you to use **slices** to access multiple indexes at the same time.
For example, if I want to access the indexes 6, 7, 8, 9, and 10.  You can use


`````python
w[6:11]
`````

[6:11] means the start position is from index 6 and the end position is index 10. For Python string slicing range, the upper-bound is excluded, which means that [6:11] actually slices the characters from 6 -> 10. The syntax for slicing in Python is [start:end:step], the 3rd one - step is optional.


You can ignore the **end** position if you want your slice to go to the end of the string. For example, the following command accesses the string from index 6 until the end. 

`````python
w[6:11:2]
`````

`````python
w[6:]
`````

You can ignore the **start** position if you want your slice to go from the begining of the string. For example

`````python
w[:5]
`````

You can also use a negative index when slicing, which means counting from the end of the string. 

For example, -1 means the position before the last character, -2 means the 2nd to last position, and so on.

`````python
#run the cell and check the output. Do these results make sense? 
w[6:], w[6:-1], w[6:-2],w[6:-3], w[6:-4]
`````
