# Iterations

Many tasks require doing the same basic actions over and over with only small changes. Since repetitive tasks appear so frequently, it is only natural that programming languages like Python would have direct methods of performing iteration.

For instance, assume that you want to check if every number in the following list is larger than zero

```Pyhton
LIST = [0,1,2,3]
```

Using our current knowledge, the only way to achieve this is to manually test each item using their index like

```Python
LIST[0]>0
.
.
.
LIST[3]>0
```

Such solution is feasible for short lists, but if the list contains hundreds of elements, it becomes a problem. 


Programming languages include basic operations that allow us to repeat some basic code while changing a counter or iterative element. 
Next, we will learn the basic ways to program iterative tasks. 

# For Loops

A for-loop repeats a set of instructions for every value in a sequence. Sometimes for-loops are referred to as definite loops because they have a predefined start and end as they are bounded by the sequence.

A for-loop is built as follows :
```python

for looping variable in sequence:
    code block

```

````python

for n in [0,1,2,3]:
    print(n)
````

WHAT IS HAPPENING?

The loop starts and assigns the variable `n` the first value on the list

Step 1.  The variable n is assigned the value 0 and printed in the screen.

Step 2.  The variable n is assigned the value 1 and printed in the screen.

Step 3.  The variable i is assigned the value 2 and printed in the screen.

Step 4.  The variable i is assigned the value 3 and printed in the screen.

With no more values to assign in the list, the for-loop is terminated. The variable `n` remains in memory with a value of n = 3.

````python
print(n)
````

Iterating over a sequence of number is very common, so python includes a built-in function, `range`, to rapidly create sequence of numbers.  

For instance, the previous code, can be adjusted as:


````python
for i in range(0, 4):
    print(i)
````

`range` works similarly to slices, it receives three inputs, a `start` indicating the starting value; a `stop` indicating the end value, which is non-inclusive, and a `step`, indicating the steps between consecutive numbers, so for instance

`range(0,4,1)` -> `0,1,2,3`

`range(0,7,2)` -> `0,2,4,6`

`range(1,9,3)` -> `1,4,7`

by default, the starting value is `0` and the step is `1`, so you can skip them

`range(4)` -> `0,1,2,3`

````python
#check if a string contains a number

def my_function(input_string = 'hi'):
    '''
    A function that checks if a string contains a number
    '''
    isanumber = False
    
    assert type(input_string)==str, 'input must be a string.'
    
    for i in input_string:
        if i.isdigit(): #this function checks if `i` is a number 
            isanumber = True
            
    return isanumber
````

````python
my_function('0eee')
````

````python
#add the numbers in a string

def my_function(input_string = 'hi'):
    
    sumofnumbers = 0
    
    assert type(input_string)==str, 'input must be a string.'
    
    for i in input_string:
        if i.isdigit():
            sumofnumbers = sumofnumbers + int(i)
            
    return sumofnumbers
````

````python
my_function('e2e2')
````

# While Loops

A while-loop is a set of instructions that is repeated, or iterated, as long a condition is true 

A while-loop is build as follows :
```python

while logical expression:
    # Code block to be repeated until logical statement is false
    code block

```

````python
n = 10

while n>1:
    
    print(n)
    
    n = n/2 
````

````python
print(n)
````

While-loops are dangerous as it is easy to create *infinite loops*. 

````python
n = 0
while n > -1:
    n += 1
````

This iteration will never end, the only way to stop the iteration is to Stop, Restart or Shutdown the kernel.

# List and dictionary comprehension*


In Python, comprehensions are an important and popular way to create lists or dictionaries. Comprehensions allow sequences to be created from other sequences with a compact syntax.

*This is a somewhat advanced topic. I'm showing this to you so that you are aware that it exists 


````python
#Example, create a list by squaring the numbers from 0 to 4

y = []
for i in range(5):
    y.append(i**2)
print(y)
````

````python
# The same result can be achieved using a compact syntax 

y = [i**2 for i in range(5)]

print(y)
````

````python
y = []
for i in range(1,5):
    for j in range(2):
        y.append(i**j)
print(y)
````

````python
y = [i**j for i in range(1,5) for j in range(2)]
print(y)
````

Similarly, we can do dictionary comprehension

````python
#create a new dictionary by finding the cube of all elements in the dictionary x
x = {'a': 1, 'b': 2, 'c': 3}

y = {key:v**3 for (key, v) in x.items()}
print(y)
````
