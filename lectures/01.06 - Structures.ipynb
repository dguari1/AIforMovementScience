## Dictionaries

A dictionary contains key-value pairs, and each key maps to a corresponding value. It is defined by using a pair of braces { }, while the elements are a list of comma-separated key: value pairs (note the key: value pair is separated by the colon, with the key at the front and value at the end).

````python
dict_1 = {'apple':3, 'orange':4, 'pear':2}
dict_1
````

To get access to a dictionary, we need to use the key of the element - dictionary[key].

````python
dict_1['apple']
````

````python
dict_1[0] # You cannot access the elements in a dictionary by index!!!!
````

You can access the keys and values of a dictionary using their methods (dictionaries are objects!!)

````python
dict_1.keys()
````

````python
dict_1.values()
````

you can fill a dictionary in multiple ways


1) Creae the dictionary using brackets 
```python
school_dict = {'UF': ['USA', 'Florida', 5], 'McGill':'Canada'}
school_dict
```

2) Create an empty dictonary fill it by defining the diferent keys and values
```python
school_dict = {}
school_dict['UF'] = ['USA', 'Florida', 5]
school_dict['McGill'] = 'Canada'
school_dict
```

3) Use the keyword ```dict``` and define a list of tuples
```python
school_dict = dict([('UF', ['USA', 'Florida', 5]), ('McGill','Canada')])
school_dict
```

you can also query if a key is in the dictionary using the keyword ```in``` or ```not in```

````python
school_dict = {'UF': 'USA', 'McGill':'Canada'}
'UF' in school_dict, 'Harvard' in school_dict, 'UC San Diego' not in school_dict
````

You could also use the list function to turn a dictionary with a list of keys. 

````python
keys = list(school_dict)
keys[0]
````

````python
school_dict[keys[1]]
````

````python
#solution to list exercise 
list_of_lists = [['apple', 3], ['banana', 10] , ['orange', 1], ['pear', 4]]
print(list_of_lists)
# Remove the item ['orange', 1]
list_of_lists.remove(['orange', 1])
print(list_of_lists)

# alternative 
list_of_lists = [['apple', 3], ['banana', 10] , ['orange', 1], ['pear', 4]]
list_of_lists.pop(2) #element in index position 2
print(list_of_lists)

#Insert the item ['mango', 5] at the end of the list
list_of_lists.append(['mango',5])
print(list_of_lists)

#Change the value of 'apple' to 6
list_of_lists[0][1]=6
print(list_of_lists)

#Insert ['coconut',1] in the index 1 of the list
list_of_lists.insert(1,['coconut',1])
print(list_of_lists)
````
